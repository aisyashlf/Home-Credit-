```python
#buat fungsi deteksi outliers
def detect_outliers_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]
    return outliers
```

```python
detect_outliers_iqr(df_app_train, 'AMT_INCOME_TOTAL')
```

```python
def find_outliers(df):
    outlier_dict = {}
    for col in df.select_dtypes(include=['int64', 'float64']).columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower) | (df[col] > upper)]
        outlier_dict[col] = len(outliers)
    return outlier_dict

find_outliers(df_app_train)
```

```python
def impute_outlier_iqr(df, col_name):
    Q1 = df[col_name].quantile(0.25)
    Q3 = df[col_name].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # outlier → NaN
    df.loc[(df[col_name] < lower) | (df[col_name] > upper), col_name] = np.nan

    # imputasi median
    df[col_name] = df[col_name].fillna(df[col_name].mode()[0]) # mode() can return multiple values, take the first
    return df

cols_impute_outlier = [
    'OWN_CAR_AGE',
    'CNT_CHILDREN',
    'CNT_FAM_MEMBERS',
    'AMT_REQ_CREDIT_BUREAU_QRT',
]

for col in cols_impute_outlier:
    if col in df_app_test.columns:
        df_app_test = impute_outlier_iqr(df_app_test, col)
```

```python
import pandas as pd

class AutoCountEncoder:
    def __init__(self, drop_original=True, cat_dtypes=('object', 'category')):
        self.drop_original = drop_original
        self.cat_dtypes = cat_dtypes
        self.cols_ = []
        self.count_maps = {}   # <── DIDEFINISIKAN DI SINI SESUAI ERROR-MU

    def fit(self, df):
        # auto detect kolom kategori
        self.cols_ = df.select_dtypes(include=self.cat_dtypes).columns.tolist()

        # buat mapping frekuensi untuk tiap kolom kategori
        for col in self.cols_:
            # cast ke object biar tidak error categorical
            series_obj = df[col].astype('object')
            self.count_maps[col] = series_obj.value_counts()

        return self

    def transform(self, df):
        df_copy = df.copy()

        for col in self.cols_:
            df_copy[col] = df_copy[col].astype('object')
            df_copy[col + "_count"] = df_copy[col].map(self.count_maps[col]).fillna(0)

        if self.drop_original:
            df_copy = df_copy.drop(columns=self.cols_)

        return df_copy

    def fit_transform(self, df):
        self.fit(df)
        return self.transform(df)
```
